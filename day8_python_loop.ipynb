# Python Loops

Loops are used to execute a block of code repeatedly.

In this notebook, we will learn:
- for loop
- while loop
- range()
- break
- continue
- pass

## 2. For Loop

```python
# Example of a for loop

for i in range(5):
    print(i)
```

### Explanation

The `for` loop is used when we want to repeat a block of code for each item in a sequence.

`range(5)` generates numbers from 0 to 4.

## 3. For Loop with a List

```python
fruits = ["Apple", "Banana", "Mango", "Orange"]

for fruit in fruits:
    print(fruit)
```

## 4. range() Function

```python
for i in range(1, 6):
    print(i)
```

```python
for i in range(2, 11, 2):
    print(i)
```

### range(start, stop, step)

- `start` → starting value
- `stop` → ending limit (not included)
- `step` → difference between values

## 5. While Loop

```python
i = 1

while i <= 5:
    print(i)
    i += 1
```

### Explanation

A `while` loop keeps running as long as the given condition is `True`.

We use `i += 1` to update the value and eventually make the condition `False`.

## 6. Break Statement

```python
for i in range(1, 11):
    if i == 6:
        break
    print(i)
```

### Explanation

`break` is used to immediately stop the loop.

## 7. Continue Statement

```python
for i in range(1, 6):
    if i == 3:
        continue
    print(i)
```

### Explanation

`continue` skips the current iteration and moves to the next iteration.

## 8. Pass Statement

```python
for i in range(5):
    pass
```

### Explanation


The `pass` statement is used as a placeholder when no action is required.

It does not stop the loop or skip an iteration.

## 9. Practical Examples

### Print even numbers

```python
for i in range(1, 21):
    if i % 2 == 0:
        print(i)
```

### Calculate sum of numbers

```python
total = 0

for i in range(1, 6):
    total += i

print("Sum:", total)
```

### Multiplication table

```python
num = 5

for i in range(1, 11):
    print(num, "x", i, "=", num * i)
```
