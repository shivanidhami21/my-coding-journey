# Python Conditional Statements(if, elif, else)

In this notebook, we will learn:
- if statement
- if-else statement
- if-elif-else statement

## 1. if Statement

```python
age = 20

if age >= 18:
    print("You can vote")
```

```python
temperature = 35

if temperature > 30:
    print("It's a hot day")
```

```python
num = 10

if num > 0:
    print("Positive Number")
```

## 2. if-else Statement

```python
age = 16

if age >= 18:
    print("You can vote")
else:
    print("You cannot vote")
```

```python
num = 7

if num % 2 == 0:
    print("Even")
else:
    print("Odd")
```

```python
password = "python123"

if password == "python123":
    print("Login Successful")
else:
    print("Wrong Password")
```

## 3. if-elif-else Statement

```python
marks = 75

if marks >= 90:
    print("Grade A")
elif marks >= 75:
    print("Grade B")
else:
    print("Grade C")
```

```python
month = 6

if month in [12, 1, 2]:
    print("Winter")
elif month in [3, 4, 5]:
    print("Summer")
else:
    print("Rainy Season")
```

```python
fruit = "mango"

if fruit == "apple":
    print("red")
elif fruit == "mango":
    print("yellow")
elif fruit == "grapes":
    print("green")
else:
    print("orange")
```

## User Input Example

```python
age = int(input("Enter your age:"))

if age >= 18:
    print("Eligible for voting")
else:
    print("Not eligible for voting")
```

```python
marks = int(input("Enter your marks:"))

if marks >= 90:
    print("Excellent")
elif marks >= 75:
    print("Good")
else:
    print("Need Improvement")
```

## Practice Example

```python
num = int(input("Enter a number:"))

if num > 0:
    print("Positive Number")
elif num < 0:
    print("Negative Number")
else:
    print("Zero")
```
