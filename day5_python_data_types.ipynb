# Data Types

## Note:
### Basic data types such as Integer, Float, String, and Boolean were covered in the Python Type Casting notebook. This focuses on collection data types: List, Tuple, Set, and Dictionary.

## 1. List

1. Ordered collection
2. Mutable (can be changed)
3. Allows duplicate values

```python
fruits = ["apple", "banana", "mango"]

print(fruits)
print(type(fruits))
```

### Used when data can change.

#### Lists are mutable, which means you can add, remove, or update items after creating the list.

```python
fruits = ["apple", "banana", "mango"]       

print("Before:", fruits)

fruits.append("orange")

print("After:", fruits)
print(type(fruits))
```

## 2. Tuple

1. Ordered collection
2. Immutable (cannot be changed)
3. Allows duplicate values

```python
num = (1, 2, 3, 4)

print(num)
print(type(num))
```

#### Tuples are immutable, which means you cannot change their items after creating them.

```python
days = ("Monday", "Tuesday", "Wednesday")

print("Before:", days)

# Trying to change the tuple
days[0] = "Sunday"

print("After:", days)
```

## 3. Set

1. Unordered collection
2. Mutable (can be changed)
3. No duplicate values allowed

```python
colors = {"red", "green", "blue"}

print(colors)
print(type(colors))
```

#### Sets are mutable. You can add or remove items, but duplicate values are not allowed.


```python
colors = {"red", "green", "blue"}    

print("Before:", colors)

colors.add("yellow")

print("After:", colors)
print(type(colors))
```

## 4. Dictionary

1. Ordered collection
2. Mutable (can be changed)
3. Stores data in key-value pairs

```python
student = {"name": "Shivani",
           "age": 21}

print(student)
print(type(student))
```

### Used to store data in key-value pairs.

#### Dictionaries are mutable. You can update existing values or add new key-value pairs.

```python
student = {"name": "Shivani",       
           "age": 21}

print("Before:", student)

student["course"] = "Data Science"

print("After:", student)
print(type(student))
```
